```python
name = "Emma James"                                         # string variable
age = 23                                                    # integer variable: no decimals
total_years_in_Stockholm = 1.5                              # float variable: number with decimals
looking_forward_to_Helsinki = True

print("My name is", name, "I am", age, "years old.")

print("Is age an integer? :", isinstance(age, int)) # Confirm if a variable is a certain type
print("What is name? : ", type(name))               # Check what type a variable is

```

```python
multiline_string = """ I speak
-Swedish
-Finnish
-English """
print(multiline_string) # use triple quotation marks for multiline string
quote = "A lyric I like is 'skynda dig älskade skynda att älska'"
print(quote) # use different kinds of quotes if string includes quote

print(" Is lyric in quote:", 'lyric' in quote) # Check to see if something is in a variable
print("Is Amos Ilonen in name:", 'Amos Ilonen' in name)
```

```python
first_name = "Emma"
surname = "James"
print("The string name has length:",len(first_name)) #Check the length of something

print("The first letter of my name is", first_name[0]) #Check what index is in a string
print("The last letter of my name is", first_name[-1]) #Check what is the last index of a string

full_name = first_name + " " + surname # string concentenation = adding strings to each other
print(full_name)
whole_name = first_name
whole_name += surname # can also add like this
print(whole_name)

# name_and_age = full_name + age: doesn't work! Name + string
name_and_age = name + "," + " "+ str(age) # concentenation works if converting an integer to a string
print(name_and_age)

repeated_name = surname * 3 # repeating strings
print(repeated_name)

name_and_age_sentence = f'My name is {name} and I am {age} years old' # String interpolation: inserting variables and expressions into a string: f-string
print(name_and_age_sentence)

print(name_and_age_sentence[11:15]) # String slicing: extracting characters from a certain index to another: non-inclusive, so won't include the index at the last character
print(name[:]) # If string slicing left blank: will return the whole thing.
print(name[0:11:2]) # Starts at index 0, ends at 10, and extracts every second character
print(name[::-1]) # Reverse a string
```
